```python
import yaml

with open(f"{dataset.location}/data.yaml", "r") as f:
    data = yaml.safe_load(f)

print(data)
```

```python
from ultralytics import YOLO

model = YOLO("yolo11n-seg.pt")  
```

```python
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=20,
    imgsz=640,
    batch=8,        
    device="cpu",   
    project="road_crack_project",
    name="yolo11_seg_crack"
)
```

```python
metrics = model.val()
print(metrics)
```
